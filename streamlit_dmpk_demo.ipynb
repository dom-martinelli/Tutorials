### The "Backend"
#### A Gentle Intro to PK Modeling

One of the main goals of PK modeling is to visualize dose clearance relationships, which of course is a function of **time**. 

We create the x-axis of the PK curve using `np.linspace`. This example gives us 100 evenly-spaced discrete points in time between 0 and 24 (hours).

```python
time = np.linspace(0, 24, 100)
```

### PK Formulae: Elimination

In real pharmacokinetics, a classic one-compartment elimination model assumes exponential decay:

$$
C(t) = C_0 \, e^{-k t}
$$

where:

- $C(t)$ is the drug concentration at time $t$
- $C_0$ is the concentration at time $t = 0$
- $k$ is the elimination rate constant (units: 1/time)

The elimination rate constant is related to clearance and volume of distribution by:

$$
k = \frac{CL}{V_d}
$$

where:

- $CL$ is clearance (e.g., L/hr)
- $V_d$ is the volume of distribution (e.g., L)
