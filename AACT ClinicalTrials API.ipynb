## PubMed ID --> NCT ID

```python
import psycopg2
import pandas as pd

pmid_list = [12515735, 20339978, 39157928]  # example subset
conn = psycopg2.connect(
    dbname="aact",
    user="your username", 
    password="your password",
    host="aact-db.ctti-clinicaltrials.org",
    port="5432",
)
pmids_str = [str(p) for p in pmid_list]

sql = """
    SELECT DISTINCT sr.nct_id
    FROM ctgov.study_references sr
    WHERE sr.pmid = ANY(%s); # 
"""

with conn, conn.cursor() as cur:
    # cur.execute() takes 2 arguments: query and variables.
    cur.execute(sql, (pmids_str,))  
    rows = cur.fetchall()

nct_ids = [r[0] for r in rows]
print(nct_ids)

nct_df = pd.DataFrame(nct_ids, columns=["nct_id"])
```
